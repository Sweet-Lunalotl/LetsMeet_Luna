```python
DDL_KUNDE = """
CREATE TABLE kunde (
    kunde_id int PRIMARY KEY,
    nachname text NOT NULL,
    vorname text NOT NULL,
    email text NOT NULL UNIQUE,
    telefon text,
    geschlecht text NOT NULL,
    geburtsdatum date NOT NULL,
    profilbild text
)
"""

DDL_INTERESSE ="""
CREATE TABLE interesse (
    interessecode_id int PRIMARY KEY,
    interessecode text NOT NULL 
)
"""

DLL_INTERESSE_PERSON ="""
CREATE TABLE interesse_person (
    kunde_id int REFERENCES kunde (kunde_id),
    interessecode_id int REFERENCES interesse (interessecode_id),
    CONSTRAINT pk_interesse_person PRIMARY KEY (kunde_id, interessecode_id)
)
"""

DDL_ADRESSE="""
CREATE TABLE adresse (
    adresse_id serial PRIMARY KEY,
    kunde_id int REFERENCES kunde (kunde_id),
    strasse_nummer text NOT NULL,
    plz text NOT NULL,
    ort text NOT NULL
)
"""

DLL_HOBBY="""
CREATE TABLE hobby (
    hobby_id int PRIMARY KEY,
    beschreibung text NOT NULL,
    quelle text
)
"""

DLL_HOBBY_KUNDE="""
CREATE TABLE hobby_kunde (
    kunde_id int REFERENCES kunde (kunde_id),
    hobby_id int REFERENCES hobby (hobby_id),
    prio integer NOT NULL,
    CONSTRAINT pk_hobby_kunde PRIMARY KEY (kunde_id, hobby_id)
)
"""

DLL_FOTO="""
CREATE TABLE foto (
    foto_id serial PRIMARY KEY,
    dateiname text NOT NULL,
    kunde_id int REFERENCES kunde (kunde_id),
    foto text NOT NULL
)
"""

DLL_NACHRICHT="""
CREATE TABLE nachricht (
    unterhaltung_id int PRIMARY KEY,
    text text NOT NULL,
    zeitpunkt timestamp NOT NULL,
    absaender int REFERENCES kunde (kunde_id),
    empfaenger int REFERENCES kunde (kunde_id)    
)
"""

DLL_LIKES="""
CREATE TABLE likes (
    person_id1 int REFERENCES kunde (kunde_id),
    person_id2 int REFERENCES kunde (kunde_id),
    status text NOT NULL,
    zeitpunkt timestamp NOT NULL
)
"""

DLL_FREUNDE="""
CREATE TABLE freunde (
    freunde_id serial PRIMARY KEY,
    person_id1 int REFERENCES kunde (kunde_id),
    person_id2 int REFERENCES kunde (kunde_id)
)
"""

with engine.begin() as conn:
    conn.execute(text("DROP TABLE IF EXISTS freunde"))
    conn.execute(text("DROP TABLE IF EXISTS likes"))
    conn.execute(text("DROP TABLE IF EXISTS nachricht"))
    conn.execute(text("DROP TABLE IF EXISTS foto"))
    conn.execute(text("DROP TABLE IF EXISTS hobby_kunde"))
    conn.execute(text("DROP TABLE IF EXISTS interesse CASCADE"))
    conn.execute(text("DROP TABLE IF EXISTS interesse_person"))
    conn.execute(text("DROP TABLE IF EXISTS adresse CASCADE"))
    conn.execute(text("DROP TABLE IF EXISTS hobby"))
    conn.execute(text("DROP TABLE IF EXISTS kunde"))
    conn.execute(text(DDL_KUNDE))
    conn.execute(text(DLL_HOBBY))
    conn.execute(text(DDL_ADRESSE))
    conn.execute(text(DDL_INTERESSE))
    conn.execute(text(DLL_INTERESSE_PERSON))
    conn.execute(text(DLL_HOBBY_KUNDE))
    conn.execute(text(DLL_FOTO))
    conn.execute(text(DLL_NACHRICHT))
    conn.execute(text(DLL_LIKES))
    conn.execute(text(DLL_FREUNDE))

print("Tabellen angelegt")
```

```python
# Hilfsfunktionen für Nachricht, Likes und Freunde
import unicodedata

def norm_email(s):
    """E-Mail vereinheitlichen (Unicode-Normalform, Leerzeichen, Groß-/Kleinschreibung)"""
    return unicodedata.normalize("NFC", str(s)).strip().lower()

# E-Mail -> kunde_id (gleiche IDs wie in der Tabelle kunde, da index_label="kunde_id" = df.index)
email_to_id = {norm_email(e): kid for kid, e in df["E-Mail"].items()}
print("E-Mails im Lookup:", len(email_to_id), "von", len(df))

def parse_ts(series):
    """Zeitstempel in beiden vorkommenden Formaten einlesen (ISO und dd.mm.yyyy)"""
    s = series.astype("string")
    iso = pd.to_datetime(s, format="%Y-%m-%d %H:%M:%S", errors="coerce")
    de = pd.to_datetime(s, format="%d.%m.%Y %H:%M:%S", errors="coerce")
    return iso.fillna(de)

def drop_incomplete(data, name):
    """Zeilen mit fehlenden Pflichtwerten (nicht auflösbare E-Mail, kaputter Zeitstempel) melden und entfernen"""
    missing = data.isna().any(axis=1)
    print(f"{name}: {missing.sum()} von {len(data)} Zeilen unvollständig")
    return data[~missing].copy()
```

```python
# Nachricht Tabelle
nachricht_data = pd.DataFrame({
    "unterhaltung_id": messages_df["conversation_id"],
    "text": messages_df["message"],
    "zeitpunkt": parse_ts(messages_df["timestamp"]),
    "absaender": messages_df["kunde_id"],
    "empfaenger": messages_df["to_email"].map(norm_email).map(email_to_id),
})
nachricht_data = drop_incomplete(nachricht_data, "nachricht")
nachricht_data[["unterhaltung_id", "absaender", "empfaenger"]] = nachricht_data[["unterhaltung_id", "absaender", "empfaenger"]].astype(int)

# Eine Unterhaltung besteht i.d.R. aus mehreren Nachrichten -> unterhaltung_id kann kein Primärschlüssel sein.
# Nur wenn die IDs nicht eindeutig sind, wird die Tabelle um einen eigenen Schlüssel erweitert.
if not nachricht_data["unterhaltung_id"].is_unique:
    print("unterhaltung_id ist nicht eindeutig -> neuer Primärschlüssel nachricht_id")
    with engine.begin() as conn:
        conn.execute(text("ALTER TABLE nachricht DROP CONSTRAINT IF EXISTS nachricht_pkey"))
        conn.execute(text("ALTER TABLE nachricht ADD COLUMN IF NOT EXISTS nachricht_id serial PRIMARY KEY"))

nachricht_data.to_sql("nachricht", engine, if_exists="append", index=False)
```

```python
# Likes Tabelle
print(likes_df["status"].value_counts())

likes_data = pd.DataFrame({
    "person_id1": likes_df["kunde_id"],
    "person_id2": likes_df["liked_email"].map(norm_email).map(email_to_id),
    "status": likes_df["status"],
    "zeitpunkt": parse_ts(likes_df["timestamp"]),
})
likes_data = drop_incomplete(likes_data, "likes")
likes_data[["person_id1", "person_id2"]] = likes_data[["person_id1", "person_id2"]].astype(int)
print("doppelte Like-Paare:", likes_data.duplicated(subset=["person_id1", "person_id2"]).sum())

likes_data.to_sql("likes", engine, if_exists="append", index=False)
```

```python
# Freunde Tabelle
# In MongoDB ist das Feld friends bei allen Nutzern leer (friends_df ist leer).
# Deshalb werden Freundschaften aus den gegenseitigen Likes (status = 'mutual') abgeleitet.
paare = []

mutual = likes_data[likes_data["status"].str.strip().str.lower() == "mutual"]
paare.append(mutual[["person_id1", "person_id2"]])

if len(friends_df) > 0:  # falls die Quelle doch Freunde enthält
    f = pd.DataFrame({
        "person_id1": friends_df["kunde_id"],
        "person_id2": friends_df["friend_email"].map(norm_email).map(email_to_id),
    })
    paare.append(drop_incomplete(f, "freunde (friends-Feld)").astype(int))

paare = pd.concat(paare, ignore_index=True)

# Jede Freundschaft nur einmal speichern (kleinere ID zuerst), keine Selbstverweise
freunde_data = pd.DataFrame({
    "person_id1": paare.min(axis=1),
    "person_id2": paare.max(axis=1),
})
freunde_data = freunde_data[freunde_data["person_id1"] != freunde_data["person_id2"]].drop_duplicates()
print("Freundschaften:", len(freunde_data))

freunde_data.to_sql("freunde", engine, if_exists="append", index=False)
```
